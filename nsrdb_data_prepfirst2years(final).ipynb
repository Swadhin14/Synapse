### Data Quality Checks for `df_2018`
We will evaluate the dataset for missing values, duplicates, and logical inconsistencies (like non-zero solar radiation at night).

```python
import pandas as pd
import numpy as np

# Re-loading data since kernel state was lost
try:
    # Attempt to load the original dataset
    # Note: If the file is not in /content/, please ensure it is uploaded.
    df = pd.read_csv('/content/NSRDB_cleaned_data.csv')
    df['datetime'] = pd.to_datetime(df['datetime'])
    df.set_index('datetime', inplace=True)

    # Re-creating the hourly 2018 slice
    df_hourly = df.resample('h').mean()
    df_2018 = df_hourly.loc['2018-01-01':'2018-12-31']

    print("--- Data Quality Checks for df_2018 ---")

    # 1. Check for missing values
    print("\n1. Missing Values:")
    print(df_2018.isnull().sum())

    # 2. Check for duplicate timestamps
    print("\n2. Duplicate Timestamps:")
    print(f"Number of duplicates in index: {df_2018.index.duplicated().sum()}")

    # 3. Check for night-time values (Solar Zenith Angle > 90 should have 0 GHI/DNI)
    night_data = df_2018[df_2018['Solar Zenith Angle'] > 95]
    non_zero_night = night_data[(night_data['GHI'] > 0) | (night_data['DNI'] > 0)]

    print("\n3. Logical Consistency (Night-time Solar):")
    print(f"Number of night-time records with non-zero solar values: {len(non_zero_night)}")

    # 4. Check for Value Ranges
    print("\n4. Value Ranges (Outlier Check):")
    display(df_2018[['Temperature', 'Pressure', 'Relative Humidity', 'GHI']].agg(['min', 'max']))

except FileNotFoundError:
    print("Error: 'NSRDB_cleaned_data.csv' not found in /content/. Please upload the file or adjust the path.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python


# --------------------------------------------------
# Load dataset
# --------------------------------------------------
df = df_2018.copy()

# --------------------------------------------------
# PV system assumptions
# --------------------------------------------------
PV_CAPACITY_KW = 100       # Installed PV capacity
SYSTEM_EFFICIENCY = 0.80   # 80% overall efficiency/derating
PANEL_TILT_DEG = 33        # Assumed fixed panel tilt

# --------------------------------------------------
# Convert angles from degrees to radians
# --------------------------------------------------
solar_zenith_rad = np.radians(df["Solar Zenith Angle"])
panel_tilt_rad = np.radians(PANEL_TILT_DEG)

# --------------------------------------------------
# Calculate beam/direct irradiance on panel
#
# Simplified assumption:
# Angle of incidence ≈ Solar Zenith Angle - Panel Tilt
# --------------------------------------------------
angle_of_incidence = solar_zenith_rad - panel_tilt_rad

beam_poa = (
    df["DNI"] * np.cos(angle_of_incidence)
)

# --------------------------------------------------
# Calculate diffuse irradiance on tilted panel
# Isotropic sky model
# --------------------------------------------------
diffuse_poa = (
    df["DHI"] * (1 + np.cos(panel_tilt_rad)) / 2
)

# --------------------------------------------------
# Total Plane-of-Array irradiance
# --------------------------------------------------
df["POA_Irradiance"] = beam_poa + diffuse_poa

# Remove physically impossible negative irradiance
df["POA_Irradiance"] = df["POA_Irradiance"].clip(lower=0)

# --------------------------------------------------
# Calculate PV output
# --------------------------------------------------
df["PV_Output_kW"] = (
    PV_CAPACITY_KW
    * (df["POA_Irradiance"] / 1000)
    * SYSTEM_EFFICIENCY
)

# Make sure output cannot exceed installed capacity
df["PV_Output_kW"] = df["PV_Output_kW"].clip(
    lower=0,
    upper=PV_CAPACITY_KW
)

# --------------------------------------------------
# Save updated dataset
# --------------------------------------------------
df.to_csv("dataset_with_pv.csv", index=False)

# --------------------------------------------------
# Check results
# --------------------------------------------------
print(
    df[
        [
            "Year",
            "Month",
            "Day",
            "Hour",
            "DNI",
            "DHI",
            "Solar Zenith Angle",
            "POA_Irradiance",
            "PV_Output_kW"
        ]
    ].head(20)
)
```
